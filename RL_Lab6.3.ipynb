Value Iteration Tmplementation

```python
import numpy as np

states = ["high", "low"]
actions = ["search", "wait", "recharge"]

V = {"high": 0.0, "low": 0.0}
gamma = 0.9
theta = 1e-6

alpha = 0.8
beta = 0.6
r_search = 2.0
r_wait = 0.5

def get_q_value(state, action):
    if state == "high":
        if action == "search":
            return alpha * (r_search + gamma * V["high"]) + (1 - alpha) * (r_search + gamma * V["low"])
        elif action == "wait":
            return r_wait + gamma * V["high"]
        elif action == "recharge":
            return 0.0 + gamma * V["high"]

    elif state == "low":
        if action == "search":
            return beta * (r_search + gamma * V["low"]) + (1 - beta) * (-3.0 + gamma * V["high"])
        elif action == "wait":
            return r_wait + gamma * V["low"]
        elif action == "recharge":
            return 0.0 + gamma * V["high"]
    return 0.0

while True:
    delta = 0
    new_V = {}
    for s in states:
        v_old = V[s]
        q_values = [get_q_value(s, a) for a in actions]
        new_V[s] = max(q_values)
        delta = max(delta, abs(v_old - new_V[s]))
    V = new_V
    if delta < theta:
        break

policy = {}
for s in states:
    q_values = {a: get_q_value(s, a) for a in actions}
    policy[s] = max(q_values, key=q_values.get)

print("Optimal Value Functions:")
for s in states:
    print(f"State '{s}': {V[s]:.4f}")

print("\nOptimal Policy:")
for s in states:
    print(f"State '{s}': Choose action '{policy[s]}'")
```
