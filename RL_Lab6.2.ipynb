Implementing Recycling Robot


```python
import numpy as np

states = ["high", "low"]
actions = ["search", "wait", "recharge"]

alpha = 0.8
beta = 0.6
r_search = 2.0
r_wait = 0.5

def transition(state, action):
    if state == "high":
        if action == "search":
            if np.random.rand() < alpha:
                return "high", r_search
            else:
                return "low", r_search
        elif action == "wait":
            return "high", r_wait
        elif action == "recharge":
            return "high", 0.0

    elif state == "low":
        if action == "search":
            if np.random.rand() < beta:
                return "low", r_search
            else:
                return "high", -3.0
        elif action == "wait":
            return "low", r_wait
        elif action == "recharge":
            return "high", 0.0

state = "high"
total_reward = 0
steps = 20

print(f"Initial State: {state}\n" + "-" * 40)

for step in range(1, steps + 1):
    if state == "high":
        action = "search"
    else:
        action = np.random.choice(["wait", "recharge"])

    next_state, reward = transition(state, action)
    total_reward += reward

    print(f"Step {step:02d} | State: {state:<4} | Action: {action:<8} | Reward: {reward:>4.1f} | Next State: {next_state}")
    state = next_state

print("-" * 40 + f"\nTotal Reward Collected: {total_reward}")
```
