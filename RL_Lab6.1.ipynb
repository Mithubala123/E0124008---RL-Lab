Implement MAB to find the best advertisement

```python
import numpy as np
import pandas as pd

df = pd.read_csv("/content/drive/MyDrive/Ads_CTR_Optimisation.csv")
N, d = df.shape

counts = np.zeros(d)
rewards = np.zeros(d)
total_reward = 0
eps = 0.1

for i in range(N):
    if np.random.rand() < eps:
        ad = np.random.randint(d)
    else:
        avg_rewards = np.where(counts > 0, rewards / counts, 0)
        ad = np.argmax(avg_rewards)

    clicked = df.iloc[i, ad]
    counts[ad] += 1
    rewards[ad] += clicked
    total_reward += clicked

best_ad = np.argmax(rewards / counts)

print(f"Total Clicks Collected: {total_reward}")
print(f"Best Advertisement Index: {best_ad}")
for idx in range(d):
    ctr = rewards[idx] / counts[idx] if counts[idx] > 0 else 0
    print(f"Ad {idx} -> Selected: {int(counts[idx])} times | CTR: {ctr:.4f}")
```
