```python
from collections import defaultdict
import numpy as np


def mc_prediction(env, policy, num_episodes, gamma=1.0):
    """First-Visit MC Prediction for estimating state-value function V."""
    # Initialize total returns and visit counts for each state
    returns_sum = defaultdict(float)
    returns_count = defaultdict(float)

    # The final value function
    V = defaultdict(float)

    for _ in range(num_episodes):
        # Generate an episode: [(state, action, reward), ...]
        episode = []
        state, _ = env.reset()
        done = False

        while not done:
            action = policy(state)
            next_state, reward, terminated, truncated, _ = env.step(action)
            done = terminated or truncated
            episode.append((state, action, reward))
            state = next_state

        # Calculate returns using first-visit approach
        visited_states = set()
        G = 0

        # Loop backwards through the episode to calculate discounted returns efficiently
        for t in reversed(range(len(episode))):
            s, a, r = episode[t]
            G = gamma * G + r

            # First-visit check
            if s not in visited_states:
                visited_states.add(s)
                returns_sum[s] += G
                returns_count[s] += 1
                V[s] = returns_sum[s] / returns_count[s]

    return V
```
